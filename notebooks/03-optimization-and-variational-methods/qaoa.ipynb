# Quantum Approximate Optimization Algorithm


```python
import numpy as np

# QAOA
from qiskit_algorithms import QAOA
from qiskit_algorithms.optimizers import *
from qiskit_algorithms.utils import algorithm_globals

# Hamiltonians / Pauli operators
from qiskit.quantum_info import Pauli, SparsePauliOp

from qiskit.primitives import StatevectorSampler

from qiskit_ibm_runtime import QiskitRuntimeService
service = QiskitRuntimeService()
```

**First, we define a function that from the coefficients of an Ising model creates the Hamiltonian for which we are going to find the ground state.**

### Hamiltonian:
### \[ H = **∑** ***Jij Zi Zj*** + **∑** ***hi Zi*** \]


```python
def get_operator(J, h, n):
    pauli_list = []

    # Coupling terms: J[i,j] Z_i Z_j
    for (i, j) in J:
        pauli_list.append(("ZZ", [i, j], J[(i, j)]))

    # Local-field terms: h[i] Z_i
    for i in h:
        pauli_list.append(("Z", [i], h[i]))

    return SparsePauliOp.from_sparse_list(
        pauli_list,
        num_qubits=n
    )
```

Now, we define the edges of the graph and obtain the Hamiltonian. For this graph, which is a cycle of length 5, the optimal solution gives a cost of -3

```python
# Edges of the graph

J1 = {(0,1):1, 
      (1,2):1, 
      (2,3):1, 
      (3,4):1, 
      (4,0):1
}

h1 = {}
n = 5

# Hamiltonian
q_op = get_operator(J1, h1, n) 

print("Hamiltonian terms:")

for pauli, coeff in q_op.to_list():
    print(f"{coeff} * {pauli}")
```

We are going to run 10 repetitions on the statevector simulator

```python
rep = 10

sampler = StatevectorSampler()
```

We run QAOA with COBYLA as the classical optimizer and with optimization level $p = 1$

```python
p = 1

val = 0.0

for i in range(rep):

    print("----- ITERATION", i, "------")

    optimizer = COBYLA()

    qaoa = QAOA(
        sampler=sampler,
        optimizer=optimizer,
        reps=p
    )

    result = qaoa.compute_minimum_eigenvalue(q_op)

    optimal_value = float(np.real(result.optimal_value))

    print("Optimal value", optimal_value)

    val += optimal_value

print("----- AVERAGE -----")

print("Average value", val / rep)
```

Now, we increase $p$ to $2$

```python
p = 2

val = 0.0

for i in range(rep):

    print("----- ITERATION", i, "------")

    optimizer = COBYLA()

    qaoa = QAOA(
        sampler=sampler,
        optimizer=optimizer,
        reps=p
    )

    result = qaoa.compute_minimum_eigenvalue(q_op)

    optimal_value = float(np.real(result.optimal_value))

    print("Optimal value", optimal_value)

    val += optimal_value

print("----- AVERAGE -----")

print("Average value", val / rep)
```

**We are going to run the algorithm with a backend which includes a noise model.**

- We didn't use the `StatevectorSampler` because 'StatevectorSampler' is appropriate for ideal local simulation and It effectively works from a pure state.

- We use `FakeOurenseV2` as the backend because IBM requires the transpiled circuit and observable to have compatible widths/layouts. `FakeOurenseV2` is a local 5-qubit snapshot of the retired Ourense device. IBM provides FakeOurenseV2 in the fake-provider module.

```python
from qiskit_ibm_runtime.fake_provider import FakeOurenseV2

from qiskit_aer import AerSimulator
from qiskit_aer.primitives import SamplerV2 as AerSampler

from qiskit.transpiler.preset_passmanagers import (
    generate_preset_pass_manager
)

rep = 10

# Modern local snapshot of the old ibmq_ourense backend
backendIBM = FakeOurenseV2()

# Local noisy Aer simulator based on Ourense
backend = AerSimulator.from_backend(backendIBM)

shots = 8192
optimization_level = 3
p = 1

# Sampler V2 running locally with Aer
sampler = AerSampler.from_backend(
    backend,
    default_shots=shots
)

# Transpile according to the 5-qubit Ourense architecture
pass_manager = generate_preset_pass_manager(
    backend=backendIBM,
    optimization_level=optimization_level
)
```

```python
p = 1

val = 0.0

for i in range(rep):

    print("----- ITERATION", i, "------")

    optimizer = COBYLA()

    qaoa = QAOA(
        sampler=sampler,
        optimizer=optimizer,
        reps=p,
        transpiler=pass_manager
    )

    result = qaoa.compute_minimum_eigenvalue(q_op)

    optimal_value = float(np.real(result.optimal_value))

    print("Optimal value", optimal_value)

    val += optimal_value

print("----- AVERAGE -----")

print("Average value", val / rep)
```
